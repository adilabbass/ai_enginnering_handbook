```python
# imports

import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
from litellm import completion
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
    
if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("Google API Key not set (and this is optional)")
```

```python
gpt_model = "openai/gpt-4o-mini"
gpt_model2 = "openai/gpt-5.6-luna"
gemini_model = "gemini/gemini-3.7-flash"

gpt_system = """Your name is Ted. You are an AI Professor and expert in LLM Engineering. You always aregument about the end of traditional softwatre enginnering.
You always keep your arugmants fact based and with references.
You are direct  but not arrogant and aggrassive. Your reply should be in 1 line.
"""
gpt2_system ="Your name is John. You are an industry expert. You always support that Software engineering and coding is still live and 100% required. You always reply based on your industry experience.  Your reply should be in 1 line"
gemini_system ="Your name is Craig. You are both professor and industry expert who lived the life of software engineer and architect, and now doing research in ground breaking AI. Your role is observer for both Ted and John. Your arguments are always based on the facts.  Your reply should be in 1 line"

gpt_user = ["Todays topic of discussion is Software development with coding is dead!"]
gpt2_user =["That's quite a topic and I donot agree"]
gemini_user =["Interesting topic. Lets start our discussion"]

```

```python
def call_gpt():
    messages=[{
        "role":"system",
        "content": gpt_system
    }]

    for gpt, gpt2 in zip(gpt_user, gpt2_user):
        messages.append({"role":"assistant", "content": gpt})
        messages.append({"role":"user", "content": gpt2 })


    response = completion(model=gpt_model, messages = messages)
    return response.choices[0].message.content
```

```python
call_gpt()
```

```python
def call_gpt2():
    messages=[{
        "role":"system",
        "content": gpt2_system
    }]

    for gpt, gpt2 in zip(gpt_user, gpt2_user):
        messages.append({"role":"user", "content": gpt})
        messages.append({"role":"assistant", "content": gpt2})
        messages.append({"role":"user", "content": gpt_user[-1] })


    response = completion(model=gpt_model2, messages = messages)
    return response.choices[0].message.content
```

```python
call_gpt2()
```

```python

display(Markdown(f"### GPT:\n{gpt_user[0]}\n"))
display(Markdown(f"### Claude:\n{gpt2_user[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_user.append(gpt_next)
    
    claude_next = call_gpt2()
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    gpt2_user.append(claude_next)
```

```python
def call_gemini():

    messages = [
        {
            "role": "system",
            "content": gemini_system
        }
    ]

    for gpt, gpt2, gemini in zip(gpt_user, gpt2_user, gemini_user):

        messages.append({
            "role": "user",
            "content": gpt
        })

        messages.append({
            "role": "user",
            "content": gpt2
        })

        messages.append({
            "role": "assistant",
            "content": gemini
        })

    # Gemini needs a new user turn before generating another response
    messages.append({
        "role": "user",
        "content": "Generate your response based on the examples above."
    })

    response = completion(
        model=gemini_model,
        messages=messages
    )

    return response.choices[0].message.content
```

```python

display(Markdown(f"### Ted:\n{gpt_user[0]}\n"))
display(Markdown(f"### John:\n{gpt2_user[0]}\n"))
display(Markdown(f"### Craig:\n{gemini_user[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### Ted:\n{gpt_next}\n"))
    gpt_user.append(gpt_next)
    
    claude_next = call_gpt2()
    display(Markdown(f"### John:\n{claude_next}\n"))
    gpt2_user.append(claude_next)

    gemini_next = call_gemini()
    display(Markdown(f"### Craig:\n{gemini_next}\n"))
    gemini_user.append(gemini_next)
```
